# Feature Engineering on Adult Dataset

## What is Feature Engineering?

### Feature engineering is the process of transforming raw data into meaningful features that improve model performance.

## Why Feature Engineering is Important
1. Improves Model Accuracy : 
    Good features help models capture real patterns.

2. Helps Handle Data Issues
    Skewness → log transform
    Sparsity → binary flags
    High categories → grouping

3. Makes Patterns Learnable
    Models struggle with:
        Non-linear relationships & Complex interactions
        Feature engineering simplifies them.

4. Reduces Overfitting
    Too many categories → overfitting
    Grouping → better generalization

5. Sometimes More Important Than Model Choice
    A simple model + good features can outperform, A complex model + poor features

## Types of Feature Engineering
1. Transformation (Changing existing features)
  Examples:
    Log transformation
    Scaling
    Normalization
    Used for: Skewed data & Linear models

2. Feature Creation (Creating new features)
    Combines or derives new meaning

3. Binning (Discretization)
    Convert continuous values into categories
    Helps: Capture non-linearity & Improve interpretability

4. Encoding (Categorical → Numeric)
  Types:
    One-hot encoding
    Label encoding
    OneHotEncoder()
   Encoding is required because models only understand numbers.

5. Feature Extraction
    Extract useful info from existing data

6. Feature Interaction
    Combine features to capture relationships

7. Feature Reduction / Grouping
    Reduce complexity
    Helps avoid: High dimensionality & Overfitting

8. Feature Selection (Closely related)
    Choose most important features
    Remove irrelevant ones (fnlwgt)
    Keep high-impact features

# Hours_Per_Week Column

```python
df['hours_per_week'].value_counts().head(10)
```

```python
df['hours_bin'] = pd.cut(df['hours_per_week'],
                        bins=[0,25,40,60,100],
                        labels=['Part-time','Full-time','Overtime','Heavy'])
df
```

```python
## Onehot Encoding Hours_per_week_bin column
hours_bin_dummies = pd.get_dummies(df['hours_bin'], prefix='hours')
df = pd.concat([df,hours_bin_dummies], axis=1)
df = df.drop('hours_bin', axis=1)
df
```

# Marital_Status Column

```python
def group_marital(x):
    if x in ['Married-civ-spouse', 'Married-AF-spouse']:
        return 'Married'
    elif x in ['Divorced', 'Separated']:
        return 'Separated'
    else:
        return 'Single'

df['marital_group'] = df['marital_status'].apply(group_marital)
```

```python
# OneHot Encoding marital_status_group column
marital_dummies = pd.get_dummies(df['marital_group'], prefix='marital')
df = pd.concat([df,marital_dummies], axis=1)
df = df.drop('marital_group', axis=1)
df = df.drop('marital_status', axis=1)
df
```

# Relationship Column: Onehot Encoding

```python
df['relationship'].value_counts()
```

```python
## Dropping original 'relationship column after Onehot Encoding
rel_dummies = pd.get_dummies(df['relationship'], prefix='rel')
df = pd.concat([df,rel_dummies], axis=1)
df = df.drop('relationship', axis=1)
df
```

# Workclass Column

```python
def group_workclass(x):
    if x == 'Private':
        return 'Private'
    elif x in ['Self-emp-not-inc', 'Self-emp-inc']:
        return 'Self-employed'
    elif x in ['Local-gov', 'State-gov', 'Federal-gov']:
        return 'Government'
    else:
        return 'Other'

df['workclass_grouped'] = df['workclass'].apply(group_workclass)
df
```

```python
df['workclass_grouped'].value_counts()
```

```python
workclass_dummies = pd.get_dummies(df['workclass_grouped'], prefix='workclass')
df = pd.concat([df,workclass_dummies], axis=1)
df = df.drop('workclass', axis=1)
df = df.drop('workclass_grouped', axis=1)
df
```

# Occupation Column

```python
def group_occupation(x):
    if x in ['Exec-managerial', 'Prof-specialty']:
        return 'White-collar'
    elif x in ['Craft-repair', 'Machine-op-inspct']:
        return 'Blue-collar'
    elif x in ['Sales', 'Adm-clerical']:
        return 'Service'
    else:
        return 'Other'

df['occupation_group'] = df['occupation'].apply(group_occupation)
```

```python
# Onehot encoding Occupation columns:
occupation_dummies = pd.get_dummies(df['occupation_group'], prefix='occupation')
df = pd.concat([df,occupation_dummies], axis=1)

# Dropping the original workclass column:
df = df.drop('occupation', axis=1)
df = df.drop('occupation_group', axis=1)
df
```

 # Native_Country

```python
df['native_country'] = df['native_country'].apply(
    lambda x: 'US' if x == 'United-States' else 'Non-US'
)
```

```python
# Binary transofrmation on native_country column

df['native_country'] = df['native_country'].map({'US': 1,'Non-US': 0})
df
```

```python
# Dropping native_country column, as to avoid bias:
df.drop(['native_country'], axis=1, inplace=True)
df
```

## Creating a new column 'Capital_Net' from capital_gain and capital_loss
### Captures hidden financial behavior
### Improves prediction for high-income group

```python
df['capital_net'] = df['capital_gain'] - df['capital_loss']

df['has_capital_activity'] = (
    (df['capital_gain'] > 0) | (df['capital_loss'] > 0)
).astype(int)

df['capital_gain_log'] = np.log1p(df['capital_gain'])
df['capital_loss_log'] = np.log1p(df['capital_loss'])
```

```python
# Dropping  redundant columns: Capital_gain & Capital_loss
df.drop(['capital_gain','capital_loss'], axis=1, inplace=True)
```

### Created 'has_capital_activity' column 
    Converts sparse data → strong signal
    Instead of: Mostly zeros (useless)
    We get:
        0 = no activity
        1 = activity
    Much easier for model to learn
### Log transformation
    1. Reduces skewness: Makes distribution more balanced
    2. Prevents dominance of outliers: Large values don’t overpower model
    3. Helps linear models: 
        Logistic Regression assumes linear relationships
        Log makes relationships smoother
