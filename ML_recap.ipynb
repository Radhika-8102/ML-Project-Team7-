# Encoding & Feature selection

## one-hot encoding

No natural order (nominal data)

e.g., color, city, gender

Number of categories is small

## label encoding

There is natural order (ordinal data)

e.g., Small < Medium < Large

Low < Medium < High

Dummy Variable Trap & multicollinearity

# feature quality / distribution check

When it is ok to drop features?

--> The ratio of the most frequent value to the second most frequent value

Rule of thumb

≈ 1 to 2 → good

> 5 or 10 → problematic

Very high → consider:

dropping feature

grouping rare categories

rebalancing

--> Too much missing values in the column

# Model Evaluation

How well your model is performing?

2. Accuracy

The simplest metric: fraction of correct predictions.

3. Precision

Of all predicted positives, how many are actually positive?

4. Recall (Sensitivity / True Positive Rate)

Of all actual positives, how many did the model correctly predict?

5. F1 Score

Mean of precision and recall
Balances precision and recall when classes are imbalanced.

Model predicted 61% correctly

6. ROC-AUC (Receiver Operating Characteristic – Area Under Curve)

Measures the model’s ability to distinguish between classes.

ROC curve plots True Positive Rate (Recall) vs False Positive Rate (FPR = FP / (FP + TN)) at different thresholds.

AUC = 1 → perfect model, 0.5 → random guessing.
